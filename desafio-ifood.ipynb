## Problema de Negócios: Análise de Conversão e "Vitrinismo" digital

Perguntas necessárias para responder a esse problema:

Qual é a correlação entre o número de visitas no mês (NumWebVisitsMonth) e o tempo desde a última compra (Recency)? Pessoas que visitam mais compram mais rápido?
